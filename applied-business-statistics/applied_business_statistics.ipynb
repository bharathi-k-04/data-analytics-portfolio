### Problem 1
A physiotherapist with a male football team is interested in studying the relationship between foot injuries and the positions at which the players play from the data collected.

Based on the above data, answer the following questions.
1.1 What is the probability that a randomly chosen player would suffer an injury?
1.2 What is the probability that a player is a forward or a winger?
1.3 What is the probability that a randomly chosen player plays in a striker position and has a foot injury?
1.4 What is the probability that a randomly chosen injured player is a striker?


```python
#Data from the table

total_players = 235

# Injured
injured = { "Striker":45,"Forward":56,"Attacking Midfielder":24,"Winger":20}

# Not Injured
not_injured = {"Striker":32,"Forward":38,"Attacking Midfielder":11,"Winger":9}

total_injured = sum(injured.values())
total_not_injured = sum(not_injured.values())

# 1.1 probability that a randomly chosen player would suffer an injury
p_injured = (total_injured / total_players)

# 1.2 Probability that a player is Forward or Winger
total_forward = injured["Forward"] + not_injured["Forward"]
total_winger = injured["Winger"] + not_injured["Winger"]
p_forward_or_winger = (total_forward + total_winger) / (total_players)

# 1.3 probability that a randomly chosen player plays in a striker position and has a foot injury
p_striker_and_injured = injured["Striker"] / total_players

# 1.4 Probability that a randomly chosen injured player is a striker
p_striker_given_injured = injured["Striker"] / total_injured

# Printing results
print("1.1 Probability of injury:", round(p_injured, 4))
print("1.2 Probability of Forward or Winger:", round(p_forward_or_winger, 4))
print("1.3 Probability of Striker AND Injured:", round(p_striker_and_injured, 4))
print("1.4 Probability of Striker given Injured:", round(p_striker_given_injured, 4))
```

#### Answers:
#### 1.1 Probability of injury: 0.617
#### 1.2 Probability of Forward or Winger: 0.5234
#### 1.3 Probability of Striker AND Injured: 0.1915
#### 1.4 Probability of Striker given Injured: 0.3103

### Problem 2
The breaking strength of gunny bags used for packaging cement is normally distributed with a mean of 5 kg per sq. centimetre and a standard deviation of 1.5 kg per sq. centimetre. The quality team of the cement company wants to know the following about the packaging material to better understand wastage or pilferage within the supply chain; Answer the questions below based on the given information; (Provide an appropriate visual representation of your answers, without which marks will be deducted)
2.1] What proportion of the gunny bags have a breaking strength of less than 3.17 kg per sq cm?
2.2] What proportion of the gunny bags have a breaking strength of at least 3.6 kg per sq cm.?
2.3] What proportion of the gunny bags have a breaking strength between 5 and 5.5 kg per sq cm.?
2.4] What proportion of the gunny bags have a breaking strength NOT between 3 and 7.5 kg per sq cm.?


```python
#Mean(μ) = 5
#Standard deviation(σ) = 1.5
#Normal Distribution
#Importing necessary libraries:
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
# 2.1 Proportion of bags with breaking strength less than 3.17

# Given parameters
mu = 5
sigma = 1.5

# probability calculation
prob_less_317 = norm.cdf(3.17, mu, sigma)
print("Proportion of bags with strength < 3.17:", round(prob_less_317, 4))
```

```python
# Create x values
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
pdf = norm.pdf(x, mu, sigma)

# Plot
plt.plot(x, pdf)
plt.axvline(3.17)
x_fill = np.linspace(mu - 4*sigma, 3.17, 500)
plt.fill_between(x_fill, norm.pdf(x_fill, mu, sigma))
plt.title("P(X < 3.17)")
plt.xlabel("Breaking Strength")
plt.ylabel("Density")
plt.show()
```

### 2.1] Answer: Proportion of bags with breaking strength less than 3.17 is 0.1112

```python
# 2.2 Proportion of bags with breaking strength of at least 3.6
# probability calculation
prob_atleast_36 = 1 - norm.cdf(3.6, mu, sigma)
print("Proportion of bags with strength ≥ 3.6:", round(prob_atleast_36, 4))
```

```python
# Create x values
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
pdf = norm.pdf(x, mu, sigma)

# Plot
plt.plot(x, pdf)
plt.axvline(3.6)
x_fill = np.linspace(3.6, mu + 4*sigma, 500)
plt.fill_between(x_fill, norm.pdf(x_fill, mu, sigma))
plt.title("P(X ≥ 3.6)")
plt.xlabel("Breaking Strength")
plt.ylabel("Density")
plt.show()
```

### 2.2] Answer: Proportion of bags with breaking strength of at least 3.6 is 0.8247

```python
# 2.3 Proportion of bags with breaking strength between 5 and 5.5

# probability calculation
prob_between_5_55 = norm.cdf(5.5, mu, sigma) - norm.cdf(5, mu, sigma)
print("Proportion of bags with strength between 5 and 5.5:", round(prob_between_5_55, 4))
```

```python
# Create x values
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
pdf = norm.pdf(x, mu, sigma)

# Plot
plt.plot(x, pdf)
plt.axvline(5)
plt.axvline(5.5)
x_fill = np.linspace(5, 5.5, 500)
plt.fill_between(x_fill, norm.pdf(x_fill, mu, sigma))
plt.title("P(5 < X < 5.5)")
plt.xlabel("Breaking Strength")
plt.ylabel("Density")
plt.show()
```

### 2.3] Answer: Proportion of bags with breaking strength between 5 and 5.5 is 0.1306

```python
# 2.4 Proportion of bags with breaking strength NOT between 3 and 7.5
# probability calculation
prob_not_between = norm.cdf(3, mu, sigma) + (1 - norm.cdf(7.5, mu, sigma))
print("Proportion of bags NOT between 3 and 7.5:", round(prob_not_between, 4))
```

```python
# Create x values
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
pdf = norm.pdf(x, mu, sigma)

# Plot
plt.plot(x, pdf)
plt.axvline(3)
plt.axvline(7.5)

# Left tail
x_left = np.linspace(mu - 4*sigma, 3, 500)
plt.fill_between(x_left, norm.pdf(x_left, mu, sigma))

# Right tail
x_right = np.linspace(7.5, mu + 4*sigma, 500)
plt.fill_between(x_right, norm.pdf(x_right, mu, sigma))

plt.title("P(X NOT between 3 and 7.5)")
plt.xlabel("Breaking Strength")
plt.ylabel("Density")
plt.show()
```

### 2.4] Answer: Proportion of bags with breaking strength not between 3 and 7.5 is 0.139

### 4.1] How does the hardness of implants vary depending on dentists?

```python
from scipy import stats
df = data.copy()

# Separate dataset by alloy
df_alloy1 = df[df['Alloy'] == 1]
df_alloy2 = df[df['Alloy'] == 2]
```

#### ANALYSIS FOR ALLOY 1:

```python
# Assumption 1: Normality (Shapiro-Wilk Test)
w, p_value = stats.shapiro(df_alloy1['Response'])
print("Shapiro-Wilk p-value (Alloy 1):", p_value)
```

```python
#Assumption 2: Equal Variance (Levene’s Test)
statistic, p_value = stats.levene(
    df_alloy1[df_alloy1['Dentist']==1]['Response'],
    df_alloy1[df_alloy1['Dentist']==2]['Response'],
    df_alloy1[df_alloy1['Dentist']==3]['Response'],
    df_alloy1[df_alloy1['Dentist']==4]['Response'],
    df_alloy1[df_alloy1['Dentist']==5]['Response']
)

print("Levene Test p-value (Alloy 1):", p_value)
```

```python
# One-Way ANOVA (Dentist Effect)
dentist1 = df_alloy1[df_alloy1['Dentist']==1]['Response']
dentist2 = df_alloy1[df_alloy1['Dentist']==2]['Response']
dentist3 = df_alloy1[df_alloy1['Dentist']==3]['Response']
dentist4 = df_alloy1[df_alloy1['Dentist']==4]['Response']
dentist5 = df_alloy1[df_alloy1['Dentist']==5]['Response']

fvalue, p_value = stats.f_oneway(dentist1, dentist2, dentist3, dentist4, dentist5)

print("ANOVA p-value (Alloy 1):", p_value)

if p_value < 0.05:
    print("Reject H0: Hardness differs among dentists.")
else:
    print("Fail to reject H0: No significant difference among dentists.")
```

#### ANALYSIS FOR ALLOY 2:

```python
# Assumption 1: Normality (Shapiro-Wilk Test)
w, p_value = stats.shapiro(df_alloy2['Response'])
print("Shapiro-Wilk p-value (Alloy 2):", p_value)
```

```python
#Assumption 2: Equal Variance (Levene’s Test)
statistic, p_value = stats.levene(
    df_alloy2[df_alloy2['Dentist']==1]['Response'],
    df_alloy2[df_alloy2['Dentist']==2]['Response'],
    df_alloy2[df_alloy2['Dentist']==3]['Response'],
    df_alloy2[df_alloy2['Dentist']==4]['Response'],
    df_alloy2[df_alloy2['Dentist']==5]['Response']
)

print("Levene Test p-value (Alloy 2):", p_value)
```

```python
# One-Way ANOVA (Dentist Effect)
dentist1 = df_alloy2[df_alloy2['Dentist']==1]['Response']
dentist2 = df_alloy2[df_alloy2['Dentist']==2]['Response']
dentist3 = df_alloy2[df_alloy2['Dentist']==3]['Response']
dentist4 = df_alloy2[df_alloy2['Dentist']==4]['Response']
dentist5 = df_alloy2[df_alloy2['Dentist']==5]['Response']

fvalue, p_value = stats.f_oneway(dentist1, dentist2, dentist3, dentist4, dentist5)

print("ANOVA p-value (Alloy 2):", p_value)

if p_value < 0.05:
    print("Reject H0: Hardness differs among dentists.")
else:
    print("Fail to reject H0: No significant difference among dentists.")
```

### 4.2] How does the hardness of implants vary depending on methods?

#### ANALYSIS FOR ALLOY 1:

```python
# Assumption 1: Normality (Shapiro-Wilk)
w, p_value = stats.shapiro(df_alloy1["Response"])
print("Shapiro p-value (Alloy 1):", p_value)

# Assumption 2: Equal Variance (Levene)
statistic, p_value = stats.levene(
    df_alloy1[df_alloy1["Method"] == 1]["Response"],
    df_alloy1[df_alloy1["Method"] == 2]["Response"],
    df_alloy1[df_alloy1["Method"] == 3]["Response"]
)
print("Levene p-value (Alloy 1):", p_value)

# Collect data for ANOVA
method1 = df_alloy1[df_alloy1["Method"] == 1]["Response"]
method2 = df_alloy1[df_alloy1["Method"] == 2]["Response"]
method3 = df_alloy1[df_alloy1["Method"] == 3]["Response"]

# One-way ANOVA
fvalue, p_value = stats.f_oneway(method1, method2, method3)
print("ANOVA p-value (Alloy 1):", p_value)

# Conclusion
if p_value < 0.05:
    print("Reject H0: Hardness differs among methods (Alloy 1).")

    
else:
    print("Fail to reject H0: No significant difference among methods (Alloy 1).")
```

#### ANALYSIS FOR ALLOY 2:

```python
# Assumption 1: Normality (Shapiro-Wilk)
w, p_value = stats.shapiro(df_alloy2["Response"])
print("Shapiro p-value (Alloy 2):", p_value)

# Assumption 2: Equal Variance (Levene)
statistic, p_value = stats.levene(
    df_alloy2[df_alloy2["Method"] == 1]["Response"],
    df_alloy2[df_alloy2["Method"] == 2]["Response"],
    df_alloy2[df_alloy2["Method"] == 3]["Response"]
)
print("Levene p-value (Alloy 2):", p_value)

# Collect data for ANOVA
method1 = df_alloy2[df_alloy2["Method"] == 1]["Response"]
method2 = df_alloy2[df_alloy2["Method"] == 2]["Response"]
method3 = df_alloy2[df_alloy2["Method"] == 3]["Response"]

# One-way ANOVA
fvalue, p_value = stats.f_oneway(method1, method2, method3)
print("ANOVA p-value (Alloy 2):", p_value)

# Conclusion
if p_value < 0.05:
    print("Reject H0: Hardness differs among methods (Alloy 2).")
else:
    print("Fail to reject H0: No significant difference among methods (Alloy 2).")
```

### 4.3] What is the interaction effect between the dentist and method on the hardness of dental implants for each type of alloy?

### 4.4] How does the hardness of implants vary depending on dentists and methods together?

```python
import statsmodels.api as sm
from statsmodels.formula.api import ols
from scipy import stats
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# ALLOY 1
df_alloy1 = df[df["Alloy"] == 1]

model1 = ols("Response ~ C(Dentist) + C(Method) + C(Dentist):C(Method)", 
             data=df_alloy1).fit()

anova_table1 = sm.stats.anova_lm(model1, typ=2)
print("\nTwo-Way ANOVA Table (Alloy 1)")
print(anova_table1)

# Assumption 1: Normality 
w, p_value = stats.shapiro(model1.resid)
print("\nShapiro p-value (Alloy 1):", p_value)

# Assumption 2: Equal Variance
groups1 = [group["Response"].values 
           for name, group in df_alloy1.groupby(["Dentist","Method"])]
stat, p_value = stats.levene(*groups1)
print("Levene p-value (Alloy 1):", p_value)


# Create interaction group
df_alloy1["Group"] = df_alloy1["Dentist"].astype(str) + "_" + df_alloy1["Method"].astype(str)

print("\nTukey Test (Alloy 1):")
tukey1 = pairwise_tukeyhsd(df_alloy1["Response"], df_alloy1["Group"])
print(tukey1)

```

```python
# ALLOY 2

df_alloy2 = df[df["Alloy"] == 2]

model2 = ols("Response ~ C(Dentist) + C(Method) + C(Dentist):C(Method)", 
             data=df_alloy2).fit()

anova_table2 = sm.stats.anova_lm(model2, typ=2)
print("\nTwo-Way ANOVA Table (Alloy 2)")
print(anova_table2)

# Assumption 1: Normality
w, p_value = stats.shapiro(model2.resid)
print("\nShapiro p-value (Alloy 2):", p_value)

# Assumption 2: Equal Variance
groups2 = [group["Response"].values 
           for name, group in df_alloy2.groupby(["Dentist","Method"])]
stat, p_value = stats.levene(*groups2)
print("Levene p-value (Alloy 2):", p_value)


df_alloy2["Group"] = df_alloy2["Dentist"].astype(str) + "_" + df_alloy2["Method"].astype(str)

print("\nTukey Test (Alloy 2):")
tukey2 = pairwise_tukeyhsd(df_alloy2["Response"], df_alloy2["Group"])
print(tukey2)
```
